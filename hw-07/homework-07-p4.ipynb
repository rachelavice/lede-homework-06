```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

response = requests.get("https://www.tnwb.uscourts.gov/Search/Search.aspx?zoom_sort=0&zoom_xml=0&zoom_query=CAR&zoom_per_page=300&zoom_and=1&zoom_cat%5B%5D=-1")
doc = BeautifulSoup(response.text, 'html.parser')
items = doc.find_all('div', attrs={"class" : ["result_block","result_altblock"]})
```

```python
len(items)
```

```python
rows=[]

for item in items[:3]:
    row = {}

    try:
        row['case_name'] = item.find('a').get_text(strip=True)     
    except:
        print("no case name found")

    try:
        row['url'] = item.find('a').get('href', None)       
    except:
        print("no url found")

    try:
        row['category'] = item.find(class_='category').text.strip()
    except:
        print("no category found")

    try:
        row['additional details'] = item.find(class_='infoline').text.strip()
    except:
        print("no additional details found")

    try:
        row['additional details'] = item.find(class_='infoline').text.strip()
    except:
        print("no additional details found")

    rows.append(row)

rows
    
```
